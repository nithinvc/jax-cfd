```python
def print_energy_spectrum_table(data, n_sampled=20):
    """
    Print energy spectrum data as a table with resolutions as columns and wavenumbers as rows.
    Samples wavenumbers uniformly in log space. Handles different spectrum lengths by using NA for missing values.
    
    Args:
        data: Dictionary with resolution keys (e.g., '512', '256') and energy spectra values
        n_sampled: Number of wavenumbers to sample uniformly in log space
    """
    # Get all resolutions and sort them numerically
    resolutions = sorted(data.keys(), key=lambda x: int(x))
    
    # Find the maximum spectrum length across all resolutions
    max_n_wavenumbers = max(len(data[res]) for res in resolutions)
    
    # Sample wavenumbers uniformly in log space (avoid k=0 for log)
    k_min = 1  # Start from k=1 to avoid log(0)
    k_max = max_n_wavenumbers - 1
    log_k_samples = np.logspace(np.log10(k_min), np.log10(k_max), n_sampled)
    wavenumber_indices = np.round(log_k_samples).astype(int)
    # Remove duplicates while preserving order
    wavenumber_indices = list(dict.fromkeys(wavenumber_indices))
    
    # Print header
    print(f"{'k':>4}", end="")
    for res in resolutions:
        print(f"{res:>12}", end="")
    print()
    
    # Print separator line
    print("-" * (4 + 12 * len(resolutions)))
    
    # Print each sampled wavenumber row
    for k in wavenumber_indices:
        print(f"{k:>4}", end="")
        for res in resolutions:
            spectrum = data[res]
            if k < len(spectrum):
                value = spectrum[k]
                if value < 1:
                    print(f"{value:>12.3e}", end="")
                else:
                    print(f"{value:>12.3f}", end="")
            else:
                print(f"{'NA':>12}", end="")
        print()
print_energy_spectrum_table(data, n_sampled=25)
```
